# POC-7 Phase 3 — Canonical Data Validation and Analytical Readiness

This notebook verifies the canonical analytical dataset used for Track A - Comparative Intelligence.

**Data version:** phase3-v2  
**Approved analytical track:** Track A - Comparative Intelligence  
**Method version:** 1.0.0

```python
from pathlib import Path
import json
import pandas as pd

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == 'notebooks':
    PROJECT_ROOT = PROJECT_ROOT.parent.parent

CANONICAL_PATH = PROJECT_ROOT / 'data' / 'canonical' / 'intelligence_data.csv'
MANIFEST_PATH = PROJECT_ROOT / 'data' / 'manifest.json'

EXPECTED_DATA_VERSION = 'phase3-v2'

print('Project root:', PROJECT_ROOT)
print('Canonical path:', CANONICAL_PATH)
```

## 1. Load canonical dataset

The canonical CSV is the mandatory analytical source of truth.

```python
df = pd.read_csv(CANONICAL_PATH)

print('Rows:', len(df))
print('Columns:', len(df.columns))
print('\nColumns:')
print(df.columns.tolist())
```

## 2. Verify canonical data version

```python
with open(MANIFEST_PATH, 'r', encoding='utf-8') as file:
    manifest = json.load(file)

manifest_version = manifest.get('data_version')
row_versions = df['data_version'].dropna().astype(str).unique().tolist()

print('Manifest data version:', manifest_version)
print('Row data versions:', row_versions)

assert manifest_version == EXPECTED_DATA_VERSION
assert row_versions == [EXPECTED_DATA_VERSION]

print('DATA VERSION CHECK: PASS')
```

## 3. Validate required canonical fields

```python
required_columns = [
    'record_id',
    'record_type',
    'observed_at',
    'entity_id',
    'category',
    'subcategory',
    'stage',
    'metric_name',
    'metric_value',
    'text_value',
    'is_synthetic',
    'data_version',
]

missing_columns = [
    column for column in required_columns
    if column not in df.columns
]

print('Missing required columns:', missing_columns)
assert not missing_columns
print('REQUIRED COLUMN CHECK: PASS')
```

## 4. Confirm canonical sample structure

```python
source_leads = df[df['record_type'] == 'crm_lead'].copy()
metric_records = df[df['record_type'] == 'crm_lead_metric'].copy()

print('Total canonical rows:', len(df))
print('Source lead records:', len(source_leads))
print('Metric records:', len(metric_records))

assert len(df) == 60
assert len(source_leads) == 30
assert len(metric_records) == 30

print('CANONICAL SAMPLE STRUCTURE: PASS')
```

## 5. Confirm synthetic-data status

```python
synthetic_values = df['is_synthetic'].dropna().astype(bool).unique().tolist()
print('Synthetic status values:', synthetic_values)
assert synthetic_values == [True]
print('SYNTHETIC STATUS CHECK: PASS')
```

## 6. Analytical readiness summary

```python
print('Analytical readiness summary')
print('----------------------------')
print('Data version:', EXPECTED_DATA_VERSION)
print('Canonical rows:', len(df))
print('Source leads:', len(source_leads))
print('Approved track: Track A - Comparative Intelligence')
print('Primary dimensions: stage, category, subcategory')
print('Interpretation: descriptive only')
print('Predictive use: not supported')
print('Causal interpretation: not supported')
print('ANALYTICAL READINESS CHECK: PASS')
```

## Conclusion

The canonical dataset is loaded from the required canonical path, the expected data version is confirmed, required analytical fields are present, the 60-row canonical structure is confirmed, and the 30-source-lead analytical sample is available for the approved Track A workflow.
